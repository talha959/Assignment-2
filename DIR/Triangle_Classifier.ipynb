```
def classify_triangle(a, b, c):
    """
    Classifies a triangle based on the lengths of its three sides.

    Args:
        a (float): Length of side a.
        b (float): Length of side b.
        c (float): Length of side c.

    Returns:
        str: A string describing the triangle type.
    """
    # --- Triangle Validity Check ---
    # The sum of the lengths of any two sides of a triangle must be greater
    # than the length of the third side.
    if a + b <= c or a + c <= b or b + c <= a:
        return "Invalid Triangle: The given sides do not form a triangle."

    # --- Classification by Side ---
    side_type = ""
    if a == b == c:
        side_type = "Equilateral"
    elif a == b or b == c or a == c:
        side_type = "Isosceles"
    else:
        side_type = "Scalene"

    # --- Classification by Angle ---
    # Using the converse of the Pythagorean theorem.
    # We must identify the longest side first.
    sides = sorted([a, b, c])
    hypotenuse_squared = sides[2]**2
    other_sides_squared_sum = sides[0]**2 + sides[1]**2

    angle_type = ""
    if abs(hypotenuse_squared - other_sides_squared_sum) < 0.00001:
        # Using a small tolerance for floating point comparison
        angle_type = "Right"
    elif hypotenuse_squared < other_sides_squared_sum:
        angle_type = "Acute"
    else:
        angle_type = "Obtuse"

    return f"Valid Triangle: It is a {side_type} and {angle_type} triangle."


# --- Main execution block ---
if __name__ == "__main__":
    try:
        print("Enter the three side lengths of the triangle:")
        side1 = float(input("Enter side 1: "))
        side2 = float(input("Enter side 2: "))
        side3 = float(input("Enter side 3: "))

        # Set a breakpoint here to debug the inputs
        if side1 <= 0 or side2 <= 0 or side3 <= 0:
             print("Error: Side lengths must be positive numbers.")
        else:
            result = classify_triangle(side1, side2, side3)
            print(result)

    except ValueError:
        print("Error: Invalid input. Please enter numbers only.")
```
